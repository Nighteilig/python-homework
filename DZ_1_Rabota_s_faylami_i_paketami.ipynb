## Работа с файлами и пакетами

### Задание 1

Перевести содержимое файла `purchase_log.txt` в словарь `purchases`.

```python
import json

purchases = {}

with open('purchase_log.txt', encoding='utf-8') as file:
    next(file)
    for line in file:
        purchase = json.loads(line)
        purchases[purchase['user_id']] = purchase['category']
```

```python
for user_id in list(purchases)[:2]:
    print(user_id, purchases[user_id])
```

### Задание 2 (необязательное)

Добавить к визитам категорию покупки и записать результат в файл `funnel.csv`. Файл `visit_log.csv` обрабатывается построчно.

```python
import csv

with open('visit_log.csv', encoding='utf-8', newline='') as visits_file, \
        open('funnel.csv', 'w', encoding='utf-8', newline='') as funnel_file:
    visits = csv.DictReader(visits_file)
    writer = csv.writer(funnel_file)
    writer.writerow(['user_id', 'source', 'category'])

    for visit in visits:
        user_id = visit['user_id']
        if user_id in purchases:
            writer.writerow([user_id, visit['source'], purchases[user_id]])
```

```python
with open('funnel.csv', encoding='utf-8') as file:
    for _ in range(3):
        print(file.readline().strip())
```
