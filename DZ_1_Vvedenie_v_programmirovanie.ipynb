## Задание 1. Високосный год

Год является високосным, если он делится на 400 либо делится на 4, но не делится на 100.

```python
year = 2020

if year % 400 == 0 or (year % 4 == 0 and year % 100 != 0):
    print("Високосный год")
else:
    print("Обычный год")
```

Проверка второго примера из условия:

```python
year = 2019

if year % 400 == 0 or (year % 4 == 0 and year % 100 != 0):
    print("Високосный год")
else:
    print("Обычный год")
```

## Задание 2 (дополнительное). Счастливый билет

Билет считается счастливым, если сумма первых трёх цифр шестизначного номера совпадает с суммой последних трёх цифр.

```python
number = 123456

first_sum = number // 100000 + number // 10000 % 10 + number // 1000 % 10
last_sum = number // 100 % 10 + number // 10 % 10 + number % 10

if first_sum == last_sum:
    print("Счастливый билет")
else:
    print("Несчастливый билет")
```

Проверка второго примера из условия:

```python
number = 123321

first_sum = number // 100000 + number // 10000 % 10 + number // 1000 % 10
last_sum = number // 100 % 10 + number // 10 % 10 + number % 10

if first_sum == last_sum:
    print("Счастливый билет")
else:
    print("Несчастливый билет")
```

- Для `number = 123321` программа выводит «Счастливый билет».
