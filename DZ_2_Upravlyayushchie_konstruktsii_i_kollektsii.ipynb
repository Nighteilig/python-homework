## Задание 1

Если длина слова нечётная, нужно вывести среднюю букву. Если длина слова чётная - две средние буквы.

```python
word = 'test'
middle = len(word) // 2

if len(word) % 2 == 0:
    print(word[middle - 1:middle + 1])
else:
    print(word[middle])
```

## Задание 2 (дополнительное)

Имена сортируются по алфавиту. Люди с одинаковыми индексами после сортировки образуют пары.

```python
boys = ['Peter', 'Alex', 'John', 'Arthur', 'Richard']
girls = ['Kate', 'Liza', 'Kira', 'Emma', 'Trisha']

if len(boys) != len(girls):
    print('Внимание, кто-то может остаться без пары!')
else:
    print('Идеальные пары:')
    for boy, girl in zip(sorted(boys), sorted(girls)):
        print(f'{boy} и {girl}')
```
